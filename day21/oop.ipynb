```python
# class   --> blue print of a car
# object --> real car
# attributes --> features of a car
# methods --> actions of a car


class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print("Woof!")

    def sit(self):
        print(f"{self.name} is sitting")
```

```python
my_dog = Dog("Buddy", 3)
my_dog.bark()
```

```python
from itertools import count


class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count +=1
        print(self.count)
    
    def decrement(self):
        self.count -=1
    
```

```python
c = Counter()
c.increment()
c.increment()
c.increment()

```

```python
# four pillers of class

# Encapsulation  -- restrict to use outside of the class  use underscore to make it private

class BankAccount:
    def __init__(self,balance):
        self.balance = balance

```

```python
class BankAccount:
    def __init__(self,balance):
        self._balance = balance
```

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposite(self,amount):
        if amount>0:
            self.__balance += amount
```

```python
b1 = BankAccount(1000)
```
