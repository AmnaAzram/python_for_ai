# Default & Keywords Arguments

```python
def greeting(user_name="Guest"):
    print("Hello",user_name)
```

```python
greeting()
```

```python
# *args --- **kwargs

def sum_all(*args):
    t=0
    for i in args:
        t +=i
    print(t)
```

```python
s = sum_all(7,9,6)
s  # tuple
```

```python
def kw_argu(**kwargs):
    print(kwargs)

kw_argu(name="amna",age="20",gender="female")
```
