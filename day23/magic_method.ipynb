```python
# concept 3 -- __repr__ (developer/debug k liya)

class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Dog: {self.name} , {self.age} year old"

    def __repr__(self):
        return f"Dog: {self.name} , {self.age} year old"
        
```

```python
d2 = Dog("buddy",3)
```

```python
repr(d2)

# jb print krty h tb __str__ wala mily ga
# jb debug krty h __repr__ mily ga
```

### Compare 2 objects

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
```

```python
p1 = Point(10,20)

p2 = Point(10,20)

print(p1)
print(p2)
```

```python
# check object memory is same or not
p1 == p2
```

```python
# check object value is same or not

class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __eq__(self,data):
        return self.x == data.x and self.y ==data.y
       
```

```python
p1 = Point(10,20)

p2 = Point(10,20)

p1 ==p2
```
