## Custom Exception

```python
class InsufficientFundError(Exception):
    "Raised when withdrawal exceeds the balance"
    pass
```

```python
def withdraw_good(balance , amount):
    if amount > balance:
        raise InsufficientFundError(f"can not withdraw {amount} current balance is {balance}")

    return balance - amount
```

```python
withdraw_good(500,200)
```

Retry Pattern

```python
import time
```

```python
def flaky_task(attempt):
    if attempt < 3:
        raise ConnectionError("network Error")

    return "success"


max_attempt = 3

for attempt in range(1,max_attempt+1):
    try:
        result = flaky_task(attempt)
        print(f"Attempt {attempt} : {result}")
        break
    except ConnectionError as e:
        print(f"Attempt {attempt} failed {e}")

        if attempt < max_attempt:
            time.sleep(1)

        else:
            print("All attempt failed, Giving up.")
```

```python
# backoff (har bar zyda kro)

for attempt in range(1,4):
    wait = 2 ** (attempt-1) #1, 2 ,4 , 8,
    time.sleep(wait)

    print(f"Attempt {attempt}, would wait {wait} on failure")
```
