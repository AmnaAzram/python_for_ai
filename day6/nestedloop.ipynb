nested loop: A nested loop is a loop placed inside the body of another loop. for every pass of the
outer loop, the inner loop runs through all of its iterations completely.


```python
for i in range(5):
    for j in range(4):
        print(i,j)
```

```python
for i in range(1,4):
    for j in range(1,4):
        print(f"{i}-{j}", end=" " )
    print()
```

```python
for i in range(4,7):
    for j in range(1,4):
        print(f"{i}*{j}" , end=" ")
    print()

```

```python
for i in range(1,5):
    print("*"*i)
```

## Match Case

```python
user_input= input("Enter command (start/stop/pause)")

match user_input:
    case "start":
        print("start")

    case "stop":
        print("stop")

    case "pause":
        print("pause")

    case _:
        print("Not Valid command")


```
