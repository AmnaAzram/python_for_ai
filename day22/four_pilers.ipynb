### Inheritance

```python
# Inheritance -- child class recevied parent features -- Parent is Base class --- child is drived class
# -- parent - base - super class
# -- child - drived - sub class

class Animal:
    def __init__(self,name):
        self.name= name

    def eat(self):
        print(f"{self.name} is eating")
        
    def sleep(self):
        print(f"{self.name} is sleeping")
        
```

```python
cow = Animal("cow")
cow.sleep()
```

```python
class Dog(Animal):
    def bark(self):
        print(f"{self.name} is barking")
```

```python
d = Dog("Tommy")
d.bark()
```

```python
class Dog(Animal):
    def __init__(self, name, breed_name):
        super().__init__(name)
        self.breed_name = breed_name

    def bark(self):
        print(f"{self.name} is barking")

    def breed(self):
        print(f"{self.name} is a {self.breed_name}")

        
```

```python
d1 = Dog("tommy", "Labrador")
d1.name
d1.breed()
```
