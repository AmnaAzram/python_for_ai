```python
for item in ['laptop','keyboard','monitor']:
    if item == 'laptop':
         print(f"Bought 3 laptop")

    elif item == 'keyboard':
        print("Bought 3 keyboard")

    elif item == 'monitor':
        print("Bought 2 keyboard")

print("end of the program")

    

   
```

```python
for num in range(0,10,2):
    print(num)
```

```python
# zip (number + items)  ==== 2 chzo ko combine krta h
items = ['laptop', 'keyboard','mouse','mointer']
item_count = [4,3,2,1]

for item_c , item in zip(item_count,items):
    print(f"{item_c} {item}")
```

```python
# enumarator
items = ['laptop', 'keyboard','mouse','mointer']

for index , item in enumerate(items):

    print(f"{index} {item}")
    print(f"{"-"*10}")
```

```python
nums = [1,2,3,4,5]

# we need  revese number

for num in reversed(nums):
    print(num,end=" ") 
```
