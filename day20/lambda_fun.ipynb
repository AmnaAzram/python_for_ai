### Lambda , map , filter

```python
# lambda function : a one line function small anonymous function

def square(x:int)->int:
    return x**2
```

```python
square(3)
```

```python
s = lambda x: x**2
s(4)
```

```python
nums = [3,4,5,6]

s = list(map(lambda x: x**2, nums))
s
```

```python
# filter
numbers = [3,5,7,1,11,34]

n = list(filter(lambda x: x>5,numbers))
n
```

```python
# use function as a argument

def apply(func, value):
    print("Executing function inside Function")
    return func(value)


```

```python
def func(x):
    return x**2
    
```

```python
apply(func,5)
```

```python
apply(lambda x: x**2,5)
```
