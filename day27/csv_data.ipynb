### CSV & Regex

`csv -- a,b,c comma seprated values`

```python
import csv


rows = [
    ["Name", "Marks"],
    ["Asha",85],
    ["Rahul",70],
]

with open("marks.csv","w", newline="",encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(rows)
```

```python
with open("marks.csv","r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in  reader:
        print(row)
  
    
```

```python
# concept_2:  csv as dictionary
with open("marks.csv","r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in  reader:
        print(f"{row["Name"]} {row["Marks"]}")
  
```

```python
# concept-3: Regex(Patterns dhoondana)
# in python re - regex

import re
```

```python
text = "My contact : email >> amnaazram2828@gmail.com"

number = re.findall(r"\d+",text)
print(number)
```

```python
number = re.search(r"\d+",text)
print(number.group())
```

```python
word = re.search(r"\w+@\w+\.\w+",text)
print(word.group())
```
