```python
# sorting
# custom sorting
# A sort key is a function passed as the key argumant to sorted() or .sort(). it is called on each element to produce 
# a value that is used  for comprison , allowing items to be ordered by a derived property rather than their value 

students = [
    {"name":"Ash", "marks":85},
    {"name":"shanzi", "marks":76},
    {"name":"john", "marks":94},
]
```

```python
sorted(students,key="marks")
```

```python
# lambda function solution
sorted(students,key=lambda s: s["marks"])
```
