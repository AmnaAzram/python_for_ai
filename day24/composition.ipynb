## Property

```concept 1-- computed property(apny ap calculate)```

````python
class Circle:
    def __init__(self,radius) -> None:
        self.radius = radius
    @property
    def area(self):
        return 3.14159 * self.radius **2
````

````python
c = Circle(4)
c.area
````

````python
class Temperature:
    def __init__(self,celsius=0) -> None:
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self,value):
        if value < -273.15:
            raise  ValueError("Below absolute zero")


        self._celsius = value
````

````python
temp = Temperature()
temp.celsius

temp.celsius = -5
temp.celsius
````

### Composition -- class k under class ko call krna

````python
class Engine:
    def __init__(self,hoursepower) -> None:
        self.hoursepower = hoursepower

    def start(self):
        return f"{self.hoursepower} Engine Started"
````

````python
Engine(4).start()
````

````python
class Car(Engine):
    def __init__(self,brand, hoursepower) -> None:
        self.brand = brand
        self.hoursepower = hoursepower
        self.engine = Engine(self.hoursepower).start()

    def start(self):
        return self.engine
       
````

````python
Car("tata",1500).start()
````
