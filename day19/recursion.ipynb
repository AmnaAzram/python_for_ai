```python
from sqlite3.dbapi2 import PARSE_COLNAMES


def countdown(n):
    print(n)
    if n ==1:
        return
    countdown(n-1)


countdown(5)
```

```python
# factorial
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)
```

```python
factorial(5)
```

```python
# nested data 

def sum_nest_list(list_data):
    total = 0

    for item in list_data:
        if isinstance(item, list):
            total = total + sum_nest_list(item)

        else:
            total = total + item
    
    return total
```

```python
sum_nest_list([3,4,[3,[3,2]],4])
```
