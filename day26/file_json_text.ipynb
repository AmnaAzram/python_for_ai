# Reading & Writing files

open() build in function to open files

```python
# write mode -- write the file
with open("my_first_file.txt","a") as file:
    text = "hello this is my first file\n"
    file.write(text)
```

```python
# read mode -- read the file
with open("my_first_file.txt","r")as f:
    data = f.read()
    print(data)
```

```python
with open("my_first_file.txt",'r') as f:
    for line in f:
        print(line.strip())
        
```

```python
# utf-8 encoding python 3.15 sported
with open("my_first_file.txt","r",encoding="utf-8") as f:
    print("hello")

    
```

```python
def add_entry(text:str)->None:
    with open("diary.txt","a") as f:
        f.write(text + "\n")


def read_diary()->str:
    with open("diary.txt","r") as f:
        return f.read()


add_entry("I am learning python with ai")
add_entry("I have completed 10 weeks ")
add_entry("Now i am learning store data in files")
```

```python
print(read_diary())
```
